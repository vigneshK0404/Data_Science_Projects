```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import re
import math
from matplotlib.colors import LogNorm
```

```python
df_raw = pd.read_csv("data/icecube_public.txt")
```

```python
# The data is structured as a N x 1 table, I am getting the column header to access the data
col_str = str(list(df_raw.columns)[0])
col_headers = col_str.split()[1:]
data_cols = ['MJD[days]','log10(E/GeV)','AngErr[deg]', 'RA[deg]', 'Dec[deg]']

#I am using a regular expression to filter only numeric strings \d+ matches with 1 or more numeric characters in a string.
df_stripped = df_raw[col_str].str.strip()
is_numeric = df_stripped.str.match(r"\d+")
df_string = df_stripped[is_numeric]
```

```python
#I am splitting the 1 column of data with another regex which matches with one ore more whitespace characters.
#I convert the table to float dtype.
df = pd.DataFrame()
df[col_headers] = df_string.str.split("\s+", expand=True)
df = df.astype(float)
```

We care about sin(declination) instead of declination because it represents a physical ratio of how much sky we see and in what direction relative to the radius of the earth. Also in pure degrees or radians, we may see effects of earth geometry in the results, there is a larger solid angle around the equator than the poles, we do not want to be biased to that.

## Question 5 - Mollewide Scatter Plot

```python
#I did not realize there was a function in matplotlib for this.
#I recursively solve for the appropriate theta from the latitude from an expression I found on wikipedia
def solve_theta(latitude, tolerance, max_steps = 30):
    error_flag = np.abs(np.abs(latitude) - np.pi/2) <= tolerance
    theta = latitude.copy()
    theta[error_flag] = 0
    theta_func = lambda x : x - ((2*x + np.sin(2*x) - np.pi*np.sin(latitude))/(2 + 2*np.cos(2*x)))

    diff = np.abs(theta_func(theta) - theta)

    steps = 0
    while((np.max(diff[~error_flag]) >= tolerance) and (steps < max_steps)):
        theta_new = theta_func(theta)
        diff = np.abs(theta_new - theta)
        theta = theta_new
        steps += 1

    theta[error_flag] = latitude[error_flag]
    return theta

#I wrote the logic myself instead of using scipy's fsolve as this was much faster and I could use the entire dataset instead of a limited one.
def mollewide_proj(longitude, latitude, R, central_meredian = 180):
    delta_longitude = (longitude - central_meredian + 180) % 360 - 180 #Converts longitude from [0,360] -> [-180,180] as the mollewide formula expects it as such
    latitude = latitude * np.pi/180
    delta_longitude *= np.pi/180
    theta_solution = solve_theta(latitude, tolerance = 1e-8)   
    x = (R*2*np.sqrt(2)/np.pi)*delta_longitude*np.cos(theta_solution)
    y = R*np.sqrt(2)*np.sin(theta_solution)
    return x,y
```

```python
x,y = mollewide_proj(df["RA[deg]"], df["Dec[deg]"], R=6371000) #Radius of earth for correct scale and units.
```

```python
plt.scatter(x,y,s=0.0001, c="red", alpha=0.3)
plt.xlabel("x [m]")
plt.ylabel("y [m]")
plt.title("Right ascension vs Declination in Mollewide Projected Space")
plt.show()
```

If we want to plot the declination againt Right Ascension as a scatter plot, we are plotting in flat space. While the units match, we can get more information if we standardize them to x,y as a projection. If the information we care about from a scatter plot is the relative density, then a equal area projection is useful.
